```python
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
fm.fontManager.addfont('font/times/times.ttf')
plt.rc('font',family='Times New Roman')
plt.rcParams.update({'font.size': 14})
import numpy as np
import seaborn as sns
```

```python
xticks=["Onehot","OPF", "PSSM", "ESM"]
yticks=[
    ["ATPB","CC","TF","Else"], #MF
    ["Nuc.","Mito.","PM","ECM"],#CC
    ["CD","ST","MP","IR"]#BP
]
data_matrix = []
data_matrix.append(np.asmatrix([
            [0.13,0.20,0.38,0.81],
            [0.07,0.24,0.35,0.91],
            [0.10,0.15,0.27,0.88],
            [0.17,0.29,0.48,0.85]
        ]))
data_matrix.append(np.asmatrix([
            [0.19,0.26,0.63,0.75],
            [0.16,0.31,0.54,0.73],
            [0.04,0.14,0.69,0.68],
            [0.12,0.25,0.62,0.87]]))  
data_matrix.append(np.asmatrix([
            [0.16,0.38,0.42,0.84],
            [0.08,0.31,0.47,0.75],
            [0.09,0.35,0.46,0.82],
            [0.11,0.39,0.53,0.81]])) 

datasets=["MFO", "CCO", "BPO"]
plt.figure(figsize=(15, 5),dpi=600)
for i in range(3):
    plt.subplot(1, 3, i+1)    
    sns.heatmap(data_matrix[i],
                    linewidth=0.5,
                    # 将具体的数字写在对应的表格中，%.1f 指定了样式，在较复杂的样式中可以去掉
                    annot=np.array(['%.2f' % point for point in np.array(data_matrix[i].ravel())[0]]).reshape(np.shape(data_matrix[i])),
                    # 这里必须置空，否则会出现问题
                    fmt='',
                    yticklabels=yticks[i],
                    # 如果 usetext=True, 这里可以使用 latex 语法比如 $\leq$ = <
                    xticklabels=xticks,
                    vmax=1,
                    vmin=0,
                    # cmap 决定了注意力图的色调
                    cmap="YlGnBu")
    plt.title(datasets[i],fontsize=18)
    plt.ylabel("GO Labels", labelpad=5, fontsize=18)
    plt.xlabel("Views", labelpad=5, fontsize=18)

# plt.subplots_adjust(wspace =0.5, hspace =0.3)#调整子图间距 (0,1)
plt.tight_layout()#调整整体空白
plt.savefig("fig.viewattetinon.svg", format = "svg")
plt.savefig("fig.viewattetinon.jpg", format = "jpg", dpi=300)
plt.show()
```
